# Data Cleaning & Preprocessing

```python
df['Inward Date'].dtypes
```

```python
df['Inward Date']   = pd.to_datetime(df['Inward Date'])
df['Dispatch Date'] = pd.to_datetime(df['Dispatch Date'])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df['Core Specification'] = df['Core Specification'].fillna('Not find')
df['SSD'] = df['SSD'].fillna('Not find')
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
df['Storage Days'] = (df['Dispatch Date']-df['Inward Date']).dt.days
```

```python
df
```

```python
df['Revenue'] = (df['Price'] * df['Quantity Sold'])
```

```python
df
```

```python
df.rename(columns = {'Core Specification':'CPU'} , inplace = True)
```

```python
df
```

```python
q3 = df['Price'].quantile(0.75)
q1 = df['Price'].quantile(0.25)
iqr = q3-q1
upper_bound = q3 + 1.5*iqr
lower_bound = q1 - 1.5*iqr

df[(df['Price']<lower_bound) | (df['Price']>upper_bound)]
```
